<h1 style = "color:green">Seaborn</h1>

```python
import numpy as np
import pandas as pd

from pandas import Series, DataFrame

import matplotlib.pyplot as plt
import seaborn as sns
```

<div style = "background-color:lightblue; color:green; font-size:20px">Seaborn is data visualization library built on top of matplotlib and closely related with pandas data structures.<br>
It is more preferred when working with pandas dataframe for statistical visualization.


<div style = "background-color:lightblue; color:green; font-size:20px">When to use which ?<br>
<ul>
    <li>Matplotlib:
        <ul>
            <li>When plotting highly customized general graphs.</li>
        </ul>
    </li>
    <li>Seaborn:
        <ul>
            <li>When working with dataframes</li>
            <li>Doing EDA</li>
            <li>Statistical visualization and beautiful charts</li>
        </ul>
    </li>
</ul>

<h2 style = "color:green">1. Line Plot

```python
data = DataFrame(np.random.randint(1,100,16).reshape(8,2), columns=['A','B'])

print(data)

sns.lineplot(x = 'A', y = 'B', data = data)  
# x and y are the columns of the dataframe which we want to plot. data is the dataframe which contains those columns
# sns.lineplot(data = data) // Index of the dataframe will be taken as x-axis and values of the dataframe will be taken as y-axis


plt.grid(True)
plt.show()
```

```python
data = sns.load_dataset('tips').head(20)
print(data.head())
```

```python
sns.lineplot(x = 'total_bill', y = 'tip', data = data, hue = 'sex', size = 12, style = 'sex', palette = 'Accent', markers = ['o','>'], legend = 'brief')

plt.title('Line Plot')
plt.show()
```
