<h1 style = "color:green">Seaborn</h1>

```python
import numpy as np
import pandas as pd

from pandas import Series, DataFrame

import matplotlib.pyplot as plt
import seaborn as sns
```

<div style = "background-color:lightblue; color:green; font-size:20px">Seaborn is data visualization library built on top of matplotlib and closely related with pandas data structures.<br>
It is more preferred when working with pandas dataframe for statistical visualization.


<div style = "background-color:lightblue; color:green; font-size:20px">When to use which ?<br>
<ul>
    <li>Matplotlib:
        <ul>
            <li>When plotting highly customized general graphs.</li>
        </ul>
    </li>
    <li>Seaborn:
        <ul>
            <li>When working with dataframes</li>
            <li>Doing EDA</li>
            <li>Statistical visualization and beautiful charts</li>
        </ul>
    </li>
</ul>

<h2 style = "color:green">1. Line Plot

```python
data = DataFrame(np.random.randint(1,100,16).reshape(8,2), columns=['A','B'])

print(data)

sns.lineplot(x = 'A', y = 'B', data = data)  
# x and y are the columns of the dataframe which we want to plot. data is the dataframe which contains those columns
# sns.lineplot(data = data) // Index of the dataframe will be taken as x-axis and values of the dataframe will be taken as y-axis


plt.grid(True)
plt.show()
```

```python
data = sns.load_dataset('tips').head(20)
print(data.head())
```

```python
sns.lineplot(x = 'total_bill', y = 'tip', data = data, hue = 'sex', size = 12, style = 'sex', palette = 'Accent', markers = ['o','>'], legend = 'brief')

plt.title('Line Plot')
plt.show()
```

<h2 style ="color:green">2. Histogram

```python
data
```

```python
bins = np.arange(0,data['total_bill'].max()+10,4)
sns.displot(data['total_bill'], bins = bins, kde = True, rug = True, color = 'g')

plt.ylabel('Frequency')
plt.xticks(bins)
plt.show()
```

<div>

<div>

<h2 style = "color:green">3. Bar Plot

```python

order_of_sex = ['Female','Male']
order_of_time = ['Lunch','Dinner']
sns.barplot(x = 'sex', y = 'total_bill', data = data, hue = 'time', order = order_of_sex, hue_order = order_of_time, errorbar = ('ci',50),
            saturation = 0.4, edgecolor='black', err_kws = {
                'color':'blue',
                'linewidth':2,
            }, alpha = 0.8, dodge = False)
#Error bar (Ci) is the confidence interval that shows uncertainty in the estimate of the mean.
plt.show()
```

<h2 style ="color:green">5. Heatmap Plot

<div style = "background-color:lightblue; color:green; font-size:20px">
<ul>
    <li>Purpose: It uses different sets of colors in 2x2 grid to show how frequent is values in the dataset.</li>
    <li>Use when:<ul>
        <li>Dataset is too large for individual plotting.</li>
        <li>To visualize magnitude of high and low values along with their frequencies.</li>
        </ul>
        </li>
</ul>
</div>

```python
np.random.seed(44)
var = np.random.randint(1,100,100).reshape(10,10)
print(var)

sns.heatmap(var, vmin = 0, vmax = 100, cmap = 'Purples', annot = True, annot_kws = {
    'fontsize':12,
    'color':'green'
}, linecolor = 'lightgreen', linewidth = 2)


plt.xlabel('Columns')
plt.ylabel('Rows')
plt.show()
```

<h2 style = "color:green">6. Count Plot

<div style = "background-color:lightblue; color:green; font-size:20px">
<ul>
    <li>Purpose: It count the number of observations by category.</li>
    <li>How is it different from barplot: Count plot always plots frequency of different category, barplot plots aggregated(sum,mean) across different category.</li>
</ul>
</div>

```python
sns.countplot( x= 'sex', data = data, hue = 'smoker', palette = 'icefire', saturation = 0.5, alpha = 0.5)

plt.show()
```

<h2 style = "color:green">7.Violin Plot

<div style = "background-color:lightblue; color:green; font-size:20px">A violin plot combines a box plot with a kernel density estimation(KDE), showing the full distribution of shape of the data.<br>
Box plots just shows the spread of data using mean, median, quartile and outlier, whereas violin plot shows the distribution look of data.

```python
data = sns.load_dataset('tips')

# sns.violinplot(x = 'day', y='total_bill', data = data, hue = 'sex',palette = 'spring', split = True) 
#  To show the distribution of total bill for each day between male and female in single violin plot

sns.violinplot(x = 'day', y='total_bill', data = data, hue = 'sex',inner = 'quart',palette = 'spring')
#by default box, quart, point, stick, None

plt.legend()
plt.show()
```

<div>

<div>

<h2 style = "color:green">8. Box Plot

```python
sns.boxplot(x = 'day', y = 'total_bill', data = data, hue = 'sex', showmeans = True, palette = 'plasma')

plt.legend()
plt.show()
```
